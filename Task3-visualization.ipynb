```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Survived')
plt.title('Number of Passengers: Survived vs Not Survived')
plt.xlabel('0 = Did Not Survive, 1 = Survived')
plt.show()
```

More passengers did not survive than survived, showing the Titanic
disaster had a high overall casualty rate.

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title('Survival Count by Gender')
plt.show()
```

This chart visually confirms the earlier finding — far more women
survived than men, while far more men did not survive than women,
matching the "women and children first" pattern.

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Survival Count by Passenger Class')
plt.show()
```

1st class passengers had the highest survival counts relative to their
total, while 3rd class had the highest number of non-survivors,
confirming that passenger class strongly influenced survival chances.

```python
plt.figure(figsize=(6,4))
sns.histplot(df['Age'].dropna(), bins=30, kde=True)
plt.title('Age Distribution of Passengers')
plt.show()
```

Most passengers were between 20-40 years old, with fewer elderly
passengers and a smaller group of children onboard.

```python
plt.figure(figsize=(8,6))
numeric_df = df.select_dtypes(include='number')
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

Survived has the strongest negative correlation with Pclass, meaning
higher class numbers (3rd class) were associated with lower survival,
reinforcing the class-based survival pattern found earlier.
